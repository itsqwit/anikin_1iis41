задание 9

```python
titanic.isnull().sum()
titanic.isnull().sum() / len(titanic) * 100

embarked_top = titanic.Embarked.describe(include=['object'])['top']
titanic['Embarked'] = titanic['Embarked'].fillna(embarked_top)
age_median = titanic.Age.describe()['50%']
titanic['Age'] = titanic['Age'].fillna(age_median)

display(titanic.duplicated(subset=['PassengerId']).sum())
titanic.drop_duplicates(subset=['PassengerId'], keep='first')
```

задание 10

```python
titanic['FamilySize'] = titanic['SibSp'] + titanic['Parch'] + 1
titanic['IsAlone'] = titanic['FamilySize'].apply(lambda x: 1 if x == 1 else 0)
titanic['AgeGroup'] = pd.cut(titanic['Age'],
                             bins=[0, 12, 18, 35, 60, 100],
                             labels=['ребёнок', 'подросток', 'молодой', 'взрослый','пожилой'])
display(titanic)
```

```python
age_analysis = titanic.groupby('AgeGroup')[['Survived', 'Fare']].agg({'Survived': 'mean', 'Fare': ['mean', 'count']})
display(age_analysis)
```

```python
titanic.pivot_table(values='Survived',
 index='AgeGroup')
```

Задание 11

```python
port_counts = titanic.groupby('Embarked')['PassengerId'].count().sort_values(ascending=False)
display(port_counts)
```

```python
embarked_stats = titanic.groupby('Embarked').agg({'Age': 'mean', 'Fare': 'mean', 'Survived': 'mean', 'Pclass': lambda x: (x == 1).mean()})
display(embarked_stats)
```

Задание 12

```python
grouped_by_familysize = titanic.groupby('FamilySize').agg(Доля_выживших=('Survived', 'mean'),
                                                          Средний_возраст=('Age', 'mean'),
                                                          Средняя_стоимость_билета=('Fare', 'mean'),
                                                          Количество_пассажиров=('PassengerId', 'count'))
display(grouped_by_familysize)
```

```python
filtered_familysize = grouped_by_familysize[grouped_by_familysize['Количество_пассажиров'] > 10]
sorted_familysize = filtered_familysize.sort_values('Доля_выживших', ascending = False)
display(sorted_familysize)
```
